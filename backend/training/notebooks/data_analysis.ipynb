# Data Analysis Notebook
Exploratory data analysis for behavior sequences

## Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

sns.set_style('whitegrid')
```

## Load Datasets

```python
# Load datasets
datasets = {
    'suicide': np.load('../datasets/suicide_sequences/'),
    'pickpocket': np.load('../datasets/pickpocket_sequences/'),
    'normal': np.load('../datasets/normal_sequences/')
}

for name, data in datasets.items():
    print(f'{name}: {data.shape}')
```

## Data Statistics

```python
# Compute statistics
for name, data in datasets.items():
    print(f'\n{name.upper()}')
    print(f'  Shape: {data.shape}')
    print(f'  Mean: {data.mean():.4f}')
    print(f'  Std: {data.std():.4f}')
    print(f'  Min: {data.min():.4f}')
    print(f'  Max: {data.max():.4f}')
```

## Visualization

```python
# Plot sample sequences
fig, axes = plt.subplots(3, 1, figsize=(12, 8))

for idx, (name, data) in enumerate(datasets.items()):
    if len(data) > 0:
        sample = data[0]
        axes[idx].plot(sample)
        axes[idx].set_title(f'Sample {name.capitalize()} Sequence')
        axes[idx].set_xlabel('Time Steps')
        axes[idx].set_ylabel('Pose Values')

plt.tight_layout()
plt.show()
```
